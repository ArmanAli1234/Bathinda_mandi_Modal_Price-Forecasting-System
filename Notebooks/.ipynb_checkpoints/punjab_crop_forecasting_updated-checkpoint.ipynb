# Punjab Crop Price Forecasting — Revised Notebook

This notebook restructures the original workflow and fixes the main issues found during review:
- chronological sorting and date-based train/validation/test splits;
- lag and rolling features grouped by Commodity, Market, Grade and Variety;
- preprocessing fitted only on training data and applied consistently through pipelines;
- model selection on validation data, with a final untouched test set;
- metrics reported in the original price units;
- recursive forecasting uses the supplied dataframe and fitted pipeline (no hidden global variables).

**Important:** `lag1`, `lag7` and `lag30` mean previous 1, 7 and 30 observations within a series, not necessarily calendar days. This distinction matters if market records have missing dates.

## 1. Imports and configuration

Update `DATA_PATH` if your clean dataset is stored elsewhere. The original notebook used `../data/proceed/clean_data.csv`.

```python
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

RANDOM_STATE = 42
TARGET = "Modal_Price"
DATE_COL = "Arrival_Date"
SERIES_COLS = ["Commodity", "Market", "Grade", "Variety"]
LAG_FEATURES = ["lag1", "lag7", "lag30", "rolling_mean7", "rolling_mean30", "rolling_std7"]

DATA_PATH_CANDIDATES = [
    Path("../data/proceed/clean_data.csv"),
    Path("data/proceed/clean_data.csv"),
    Path("data/clean_data.csv"),
]
DATA_PATH = next((p for p in DATA_PATH_CANDIDATES if p.exists()), None)
if DATA_PATH is None:
    raise FileNotFoundError(
        "Could not find clean_data.csv. Set DATA_PATH manually to your dataset location."
    )
print("Using dataset:", DATA_PATH.resolve())
```

## 2. Load and validate the clean dataset

```python
df = pd.read_csv(DATA_PATH)

required_cols = [DATE_COL, TARGET, "Commodity", "Market"]
missing = [col for col in required_cols if col not in df.columns]
if missing:
    raise ValueError(f"Dataset is missing required columns: {missing}")

df[DATE_COL] = pd.to_datetime(df[DATE_COL], errors="coerce")
df[TARGET] = pd.to_numeric(df[TARGET], errors="coerce")
df = df.dropna(subset=[DATE_COL, TARGET, "Commodity", "Market"]).copy()

# Keep optional series identifiers stable even if the dataset lacks them.
for col in ["Grade", "Variety"]:
    if col not in df.columns:
        df[col] = "Unknown"
    df[col] = df[col].fillna("Unknown").astype(str)

df["Commodity"] = df["Commodity"].astype(str)
df["Market"] = df["Market"].astype(str)

# If multiple records exist for the same series/date, aggregate the target.
# This gives one observation per series per arrival date.
df = (
    df.groupby(SERIES_COLS + [DATE_COL], as_index=False, dropna=False)[TARGET]
      .mean()
      .sort_values(SERIES_COLS + [DATE_COL])
      .reset_index(drop=True)
)

print("Rows:", len(df))
print("Date range:", df[DATE_COL].min(), "to", df[DATE_COL].max())
print("Series:", df[SERIES_COLS].drop_duplicates().shape[0])
display(df.head())
```

## 3. Exploratory time-series checks (optional)

ADF is run separately for each market. It is a diagnostic only: stationarity is not a prerequisite for every tree-based regression model, and differencing is not automatically applied to the forecasting target.

```python
from statsmodels.tsa.stattools import adfuller

adf_rows = []
for market, part in df.groupby("Market"):
    series = part.sort_values(DATE_COL).groupby(DATE_COL)[TARGET].mean().dropna()
    if len(series) < 20 or series.nunique() < 2:
        adf_rows.append({"Market": market, "P_value": np.nan, "Result": "Insufficient/constant data"})
        continue
    try:
        p_value = adfuller(series)[1]
        adf_rows.append({
            "Market": market,
            "P_value": p_value,
            "Result": "Stationary at 5%" if p_value <= 0.05 else "Non-stationary at 5%"
        })
    except ValueError as exc:
        adf_rows.append({"Market": market, "P_value": np.nan, "Result": str(exc)})

adf_report = pd.DataFrame(adf_rows).sort_values("Market")
display(adf_report)
```

```python
# Plot a few market-level series rather than producing dozens of charts.
for market in df["Market"].drop_duplicates().head(5):
    part = df[df["Market"] == market].groupby(DATE_COL)[TARGET].mean().sort_index()
    plt.figure(figsize=(12, 4))
    plt.plot(part.index, part.values, label="Daily mean modal price")
    plt.plot(part.rolling(7).mean().index, part.rolling(7).mean().values,
             label="7-observation rolling mean")
    plt.title(f"Modal price trend — {market}")
    plt.xlabel("Arrival date")
    plt.ylabel("Modal price")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

## 4. Feature engineering

All lag/rolling features are shifted by one observation so the current target is not used to predict itself. Series are separated by Commodity, Market, Grade and Variety to prevent one grade/variety from contaminating another.

```python
def get_season(date):
    month = pd.Timestamp(date).month
    if month in [12, 1, 2]:
        return "Winter"
    if month in [3, 4, 5]:
        return "Summer"
    if month in [6, 7, 8, 9]:
        return "Monsoon"
    return "Autumn"


def make_features(raw_df):
    data = raw_df.copy()
    data[DATE_COL] = pd.to_datetime(data[DATE_COL], errors="coerce")
    data[TARGET] = pd.to_numeric(data[TARGET], errors="coerce")
    data = data.dropna(subset=[DATE_COL, TARGET]).copy()

    for col in SERIES_COLS:
        if col not in data.columns:
            data[col] = "Unknown"
        data[col] = data[col].fillna("Unknown").astype(str)

    data = data.sort_values(SERIES_COLS + [DATE_COL]).reset_index(drop=True)
    grouped = data.groupby(SERIES_COLS, sort=False, dropna=False)[TARGET]

    data["Year"] = data[DATE_COL].dt.year
    data["Month"] = data[DATE_COL].dt.month
    data["Day_of_week"] = data[DATE_COL].dt.dayofweek
    data["Season"] = data[DATE_COL].map(get_season)

    data["lag1"] = grouped.shift(1)
    data["lag7"] = grouped.shift(7)
    data["lag30"] = grouped.shift(30)
    data["rolling_mean7"] = grouped.transform(lambda s: s.shift(1).rolling(7, min_periods=7).mean())
    data["rolling_mean30"] = grouped.transform(lambda s: s.shift(1).rolling(30, min_periods=30).mean())
    data["rolling_std7"] = grouped.transform(lambda s: s.shift(1).rolling(7, min_periods=7).std())

    return data

featured_df = make_features(df)
feature_cols = SERIES_COLS + [
    "Year", "Month", "Day_of_week", "Season",
    "lag1", "lag7", "lag30", "rolling_mean7", "rolling_mean30", "rolling_std7"
]
model_df = featured_df.dropna(subset=feature_cols + [TARGET]).copy()
model_df = model_df.sort_values(DATE_COL).reset_index(drop=True)

print("Rows after lag feature warm-up:", len(model_df))
print("Feature columns:", feature_cols)
display(model_df[feature_cols + [TARGET, DATE_COL]].head())
```

## 5. Date-based train / validation / test split

The split is based on unique dates, not row positions after grouping. Thus, observations from the same arrival date cannot be split across train and test. Validation is used to choose the model; the test set is held back for the final unbiased check.

```python
unique_dates = np.array(sorted(model_df[DATE_COL].dropna().unique()))
if len(unique_dates) < 10:
    raise ValueError("Not enough unique dates for a reliable chronological train/validation/test split.")

train_cut = unique_dates[int(len(unique_dates) * 0.70)]
val_cut = unique_dates[int(len(unique_dates) * 0.85)]

train_df = model_df[model_df[DATE_COL] <= train_cut].copy()
val_df = model_df[(model_df[DATE_COL] > train_cut) & (model_df[DATE_COL] <= val_cut)].copy()
test_df = model_df[model_df[DATE_COL] > val_cut].copy()

if min(len(train_df), len(val_df), len(test_df)) == 0:
    raise ValueError("A chronological split produced an empty partition. Check dataset size/date coverage.")

X_train, y_train = train_df[feature_cols], train_df[TARGET]
X_val, y_val = val_df[feature_cols], val_df[TARGET]
X_test, y_test = test_df[feature_cols], test_df[TARGET]

print(f"Train: {len(train_df):,} rows through {train_df[DATE_COL].max().date()}")
print(f"Validation: {len(val_df):,} rows through {val_df[DATE_COL].max().date()}")
print(f"Test: {len(test_df):,} rows from {test_df[DATE_COL].min().date()}")
```

## 6. Preprocessing and model pipelines

Categorical encoding and numeric scaling are fitted inside each pipeline on training data only. Target scaling is handled by `TransformedTargetRegressor`, so predictions and evaluation metrics are returned in original price units.

```python
numeric_features = ["Year", "Month", "Day_of_week"] + LAG_FEATURES
categorical_features = SERIES_COLS + ["Season"]

def build_preprocessor(scale_numeric=True):
    numeric_steps = [("imputer", SimpleImputer(strategy="median"))]
    if scale_numeric:
        numeric_steps.append(("scaler", StandardScaler()))
    numeric_pipeline = Pipeline(numeric_steps)

    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ])

    return ColumnTransformer([
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features)
    ])

def build_model_pipeline(estimator, scale_numeric=True):
    regressor = Pipeline([
        ("preprocessor", build_preprocessor(scale_numeric=scale_numeric)),
        ("model", estimator)
    ])
    return TransformedTargetRegressor(
        regressor=regressor,
        transformer=StandardScaler()
    )
```

## 7. Train candidate models and select using validation MAE

```python
candidates = {
    "SGDRegressor": build_model_pipeline(
        SGDRegressor(
            loss="squared_error", penalty="l2", alpha=0.0001,
            max_iter=20000, tol=1e-5, random_state=RANDOM_STATE
        ),
        scale_numeric=True
    ),
    "RandomForest": build_model_pipeline(
        RandomForestRegressor(
            n_estimators=400, max_depth=15, min_samples_split=5,
            min_samples_leaf=2, max_features=0.75,
            random_state=RANDOM_STATE, n_jobs=-1
        ),
        scale_numeric=False
    ),
    "ExtraTrees": build_model_pipeline(
        ExtraTreesRegressor(
            n_estimators=400, max_depth=None, min_samples_leaf=2,
            max_features=0.9, random_state=RANDOM_STATE, n_jobs=-1
        ),
        scale_numeric=False
    )
}

# Add XGBoost only when the package is installed.
try:
    from xgboost import XGBRegressor
    candidates["XGBoost"] = build_model_pipeline(
        XGBRegressor(
            n_estimators=700, learning_rate=0.03, max_depth=5,
            min_child_weight=3, subsample=0.8, colsample_bytree=0.8,
            reg_alpha=0.1, reg_lambda=1.0, objective="reg:squarederror",
            eval_metric="rmse", random_state=RANDOM_STATE, n_jobs=-1
        ),
        scale_numeric=False
    )
except ImportError:
    print("xgboost is not installed; skipping XGBoost candidate.")

validation_results = []
fitted_candidates = {}

for name, pipeline in candidates.items():
    pipeline.fit(X_train, y_train)
    val_pred = pipeline.predict(X_val)
    fitted_candidates[name] = pipeline
    validation_results.append({
        "Model": name,
        "Validation_MAE": mean_absolute_error(y_val, val_pred),
        "Validation_RMSE": np.sqrt(mean_squared_error(y_val, val_pred)),
        "Validation_R2": r2_score(y_val, val_pred)
    })

validation_report = pd.DataFrame(validation_results).sort_values("Validation_MAE")
display(validation_report)
best_name = validation_report.iloc[0]["Model"]
print("Selected using validation MAE:", best_name)
```

## 8. Refit selected model and evaluate once on the untouched test set

After model selection, refit the chosen model on train + validation. The test set is not used for model selection or tuning.

```python
train_val_df = pd.concat([train_df, val_df], ignore_index=True)
X_train_val = train_val_df[feature_cols]
y_train_val = train_val_df[TARGET]

final_model = candidates[best_name]
final_model.fit(X_train_val, y_train_val)

test_pred = final_model.predict(X_test)

def regression_metrics(y_true, y_pred):
    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "MSE": mean_squared_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R2": r2_score(y_true, y_pred)
    }

test_metrics = pd.DataFrame([regression_metrics(y_test, test_pred)], index=[best_name])
display(test_metrics)
```

```python
# Actual vs predicted over the test period
comparison = test_df[[DATE_COL] + SERIES_COLS].copy()
comparison["Actual_Price"] = y_test.to_numpy()
comparison["Predicted_Price"] = test_pred
comparison = comparison.sort_values(DATE_COL)

plt.figure(figsize=(12, 5))
plt.plot(comparison[DATE_COL], comparison["Actual_Price"], label="Actual", alpha=0.8)
plt.plot(comparison[DATE_COL], comparison["Predicted_Price"], label="Predicted", alpha=0.8)
plt.title(f"Actual vs predicted — {best_name} (test set)")
plt.xlabel("Arrival date")
plt.ylabel("Modal price")
plt.legend()
plt.tight_layout()
plt.show()

display(comparison.head())
```

## 9. Recursive future forecasting

Forecasting is performed for one complete series (Commodity + Market + Grade + Variety). If Grade or Variety is omitted but multiple matching series exist, the function raises an error instead of silently mixing them.

The function starts after the latest available date for the selected series. `lag7` means the seventh previous observation, not exactly seven calendar days.

```python
def forecast_price(
    history_df,
    fitted_pipeline,
    commodity,
    market,
    start_date,
    end_date,
    grade=None,
    variety=None,
    target=TARGET
):
    data = history_df.copy()
    data[DATE_COL] = pd.to_datetime(data[DATE_COL], errors="coerce")
    data[target] = pd.to_numeric(data[target], errors="coerce")
    data = data.dropna(subset=[DATE_COL, target]).copy()

    filters = {"Commodity": commodity, "Market": market}
    if grade is not None:
        filters["Grade"] = grade
    if variety is not None:
        filters["Variety"] = variety

    for col, value in filters.items():
        data = data[data[col].astype(str) == str(value)]

    if data.empty:
        raise ValueError("No historical rows found for the requested commodity/market/grade/variety.")

    # Ensure exactly one series remains.
    remaining_series = data[SERIES_COLS].drop_duplicates()
    if len(remaining_series) != 1:
        raise ValueError(
            "More than one Grade/Variety series matches. Pass grade= and variety= explicitly. "
            f"Matching series: {remaining_series.to_dict('records')[:10]}"
        )

    data = data.sort_values(DATE_COL).reset_index(drop=True)
    if len(data) < 30:
        raise ValueError("At least 30 historical observations are required for lag30.")

    # Aggregate duplicate dates for the selected series.
    data = data.groupby(DATE_COL, as_index=False)[target].mean().sort_values(DATE_COL)
    history = data[target].astype(float).tolist()
    last_date = pd.Timestamp(data[DATE_COL].max())
    start_date = pd.Timestamp(start_date)
    end_date = pd.Timestamp(end_date)

    if start_date <= last_date:
        raise ValueError(f"start_date must be after the last historical date ({last_date.date()}).")
    if end_date < start_date:
        raise ValueError("end_date must be on or after start_date.")

    series_values = remaining_series.iloc[0].to_dict()
    forecast_dates = pd.date_range(last_date + pd.Timedelta(days=1), end_date, freq="D")
    output = []

    for forecast_date in forecast_dates:
        row = {
            **series_values,
            "Year": forecast_date.year,
            "Month": forecast_date.month,
            "Day_of_week": forecast_date.dayofweek,
            "Season": get_season(forecast_date),
            "lag1": history[-1],
            "lag7": history[-7],
            "lag30": history[-30],
            "rolling_mean7": float(np.mean(history[-7:])),
            "rolling_mean30": float(np.mean(history[-30:])),
            "rolling_std7": float(np.std(history[-7:], ddof=1))
        }
        X_future = pd.DataFrame([row], columns=feature_cols)
        predicted_price = float(fitted_pipeline.predict(X_future)[0])
        history.append(predicted_price)

        if forecast_date >= start_date:
            output.append({
                "Forecast_Date": forecast_date,
                **series_values,
                "Predicted_Price": predicted_price,
                "Season": row["Season"]
            })

    return pd.DataFrame(output)
```

## 10. Generate a forecast

Edit the values below to match a series in your dataset. Do not hard-code a last historical date; the function checks it from the selected series.

```python
# Example: update these values to a series actually present in your dataset.
FORECAST_COMMODITY = "Apple"
FORECAST_MARKET = "Bathinda APMC"
FORECAST_GRADE = None       # e.g. "Grade B"; None works only if one grade matches
FORECAST_VARIETY = None     # e.g. "Other"; None works only if one variety matches

latest_available = df[
    (df["Commodity"] == FORECAST_COMMODITY) &
    (df["Market"] == FORECAST_MARKET)
]
if FORECAST_GRADE is not None:
    latest_available = latest_available[latest_available["Grade"] == FORECAST_GRADE]
if FORECAST_VARIETY is not None:
    latest_available = latest_available[latest_available["Variety"] == FORECAST_VARIETY]

if latest_available.empty:
    print("No matching series. Check commodity/market/grade/variety values.")
else:
    last_date = latest_available[DATE_COL].max()
    forecast = forecast_price(
        history_df=df,
        fitted_pipeline=final_model,
        commodity=FORECAST_COMMODITY,
        market=FORECAST_MARKET,
        grade=FORECAST_GRADE,
        variety=FORECAST_VARIETY,
        start_date=last_date + pd.Timedelta(days=1),
        end_date=last_date + pd.Timedelta(days=30)
    )
    display(forecast.head(10))
```

## 11. Save forecast and model

The model artifact contains the complete preprocessing + estimator + target scaler. Keep the same feature definitions when loading it later.

```python
from pathlib import Path
import joblib

OUTPUT_DIR = Path("artifacts")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

joblib.dump(
    {
        "model": final_model,
        "model_name": best_name,
        "feature_columns": feature_cols,
        "series_columns": SERIES_COLS,
        "target": TARGET
    },
    OUTPUT_DIR / "punjab_crop_price_model.joblib"
)
print("Saved:", (OUTPUT_DIR / "punjab_crop_price_model.joblib").resolve())

if "forecast" in globals() and not forecast.empty:
    forecast.to_csv(OUTPUT_DIR / "crop_price_forecast.csv", index=False)
    print("Saved forecast:", (OUTPUT_DIR / "crop_price_forecast.csv").resolve())
```

## 12. Final checks / interpretation

- Compare test MAE/RMSE against a simple baseline (for example, predict the latest known price for each series) before claiming the ML model adds value.
- This notebook's lag windows are observation-based. If you need true calendar-day lags, reindex each series to daily frequency and define a missing-price policy first.
- If the dataset contains revised historical prices or future-derived columns, investigate them before training.
- Model metrics can vary substantially by commodity and market. Consider per-series evaluation, not only one pooled score.
- Recursive multi-day forecasts can drift because each prediction becomes an input to later predictions.
